```python
import numpy as np
import torch
from matplotlib import pyplot as plt
import os
import pandas as pd
```

```python
os.chdir('/u/operezc/ISSP/HADES_Project_Tests')
root_dir=os.getcwd()
print(root_dir)
```

```python
def plot_typical_loss(model_version,hpt,anomalies,version,epochs_max,colors,labels,training,validation,scaling=False):
    global root_dir
    pathor=f'{root_dir}/logs/CNN/Model-{model_version}/{hpt}_hpt/{anomalies}_poa/'
    data=[]
    fullpath=pathor+f"version_{version}/metrics.csv"

    df = pd.read_csv(fullpath)
    
    # Clean data - replace NaN with 0 and filter epochs
    df = df.fillna(0)
    df = df[df['epoch'] < epochs_max]
    
    epochs = df['epoch'].values
    cond = len(epochs)
    
    # Get accuracies from named columns
    try:
        Tacc = df['train_loss_epoch'].values[:cond]
    except:
        Tacc = df['train_loss'].values[:cond]
    Vacc = df['val_loss'].values[:cond]
    if scaling==False:
        scalefactor=1
    else:
        scalefactor=scaling
    if training==True:
        plt.plot(epochs[Tacc!=0],Tacc[Tacc!=0]*scalefactor,label=labels[0],c=colors[0])
    else:
        plt.plot(epochs[Tacc!=0],Tacc[Tacc!=0]*scalefactor,c=colors[0])
    if validation==True:
        plt.plot(epochs[Vacc!=0],Vacc[Vacc!=0]*scalefactor,label=labels[1],c=colors[1])
    else:
        plt.plot(epochs[Vacc!=0],Vacc[Vacc!=0]*scalefactor,c=colors[1])
```

```python
def plot_typical_acc(model_version,hpt,anomalies,version,epochs_max,colors,labels,training,validation,scaling=False):
    global root_dir
    pathor=f'{root_dir}/logs/CNN/Model-{model_version}/{hpt}_hpt/{anomalies}_poa/'
    data=[]
    fullpath=pathor+f"version_{version}/metrics.csv"

    df = pd.read_csv(fullpath)
    
    # Clean data - replace NaN with 0 and filter epochs
    df = df.fillna(0)
    df = df[df['epoch'] < epochs_max]
    
    epochs = df['epoch'].values
    cond = len(epochs)
    
    # Get accuracies from named columns
    try:
        Tacc = df['train_acc_epoch'].values[:cond]
    except:
        Tacc = df['train_acc'].values[:cond]
    Vacc = df['val_acc'].values[:cond]
    if scaling==False:
        scalefactor=1
    else:
        scalefactor=scaling
    if training==True:
        plt.plot(epochs[Tacc!=0],Tacc[Tacc!=0]*scalefactor,label=labels[0],c=colors[0])
    else:
        plt.plot(epochs[Tacc!=0],Tacc[Tacc!=0]*scalefactor,c=colors[0])
    if validation==True:
        plt.plot(epochs[Vacc!=0],Vacc[Vacc!=0]*scalefactor,label=labels[1],c=colors[1])
    else:
        plt.plot(epochs[Vacc!=0],Vacc[Vacc!=0]*scalefactor,c=colors[1])
```

```python
labels=["Training","Validation"]
plot_typical_acc(1,50000,0.3,0,20,["red","blue"],labels,True,True)
plt.xlabel("Epochs")
plt.ylabel("Accuracy (%)")
plt.legend()

plt.show()

labels=["Training","Validation"]

plot_typical_loss(1,50000,0.3,0,20,["red","blue"],labels,True,True)
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()

plt.show()
```
